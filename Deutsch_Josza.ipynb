```python
from qiskit import QuantumCircuit

# Our oracle XORs the output of a function with m input bits and one output bit onto one extra qubit
# We can have three functions: Constant zero, Constant one, or balanced with some arbitrary arrangement of swap gates
m = 5

# Constant one gate
# We output one if the bottom qubit was in the zero state (0 ^ 1 = 1) and zero if the bottom qubit was in the one state (1 ^ 1 = 0)
# Therefore we simply flip the bottom qubit
qc1 = QuantumCircuit(m + 1)
qc1.x(m)
gate_constant_one_oracle = qc1.to_gate()

# Constant zero gate
# If the bottom qubit is in state zero, 0 ^ 0 = 0, so we do nothing. If the bottom qubit is in state one, 1 ^ 0 = 1, so we do nothing. Therefore this is just an identity gate.
qc2 = QuantumCircuit(m + 1)
qc2.id(0)
gate_constant_one_oracle = qc2.to_gate()

# Balanced Function
# Since half the input states have a zero in any given place, we can simply pick a qubit on which to control a CNOT gate, flipping the bottom qubit for half the inputs
# This gate can be modified to fit any balanced function given appropriate swap gates
qc3 = QuantumCircuit(m + 1)
qc3.cx(0, m)
gate_balanced_oracle = qc3.to_gate()
```
